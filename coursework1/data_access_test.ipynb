```python
from astroquery.esa.euclid import Euclid

print(Euclid.get_status_messages())
Euclid.login()
```

```python
tile_id = 102159484

# Inspect table names
tables = Euclid.load_tables(only_names=True, include_shared_tables=True)
for table in tables:
    print(table.name)
```

```python
# Inspect columns for potentially useful tables
for table in [
    "q1.mer_catalogue",
    "q1.mer_cutouts",
    "q1.mer_morphology",
    "q1.mer_final_catalogue",
    "q1.observation_mosaic",
    "q1.mosaic_product",
    "q1.mer_final_catalogue",
    "q1.phz_photo_z",
    "q1.phz_classification",
    "q1.phz_catalogue",
    "q1.spe_catalogue"    
]:
    query = f"SELECT TOP 1 * FROM {table}"
    tbl = Euclid.launch_job(query).get_results()
    print("\n", table)
    print(tbl.colnames)
```

## Mosaics

```python
# Query just the tile wanted (for mosaics)
q=f"SELECT * FROM q1.mosaic_product WHERE tile_index = {tile_id}"
tbl = Euclid.launch_job(q).get_results()
tbl
```

```python
help(Euclid.get_product)
```

```python
# Get just NISP and VIS rows
tbl = tbl[(tbl['instrument_name']=='NISP') | (tbl['instrument_name']=='VIS')]

# Download product for one to test
file_name = tbl['file_name'][0]
release_name = tbl['release_name'][0]
path = Euclid.get_product(file_name=file_name, schema=release_name)
path
```

```python
from astropy.io import fits

# Check file is indeed a fits file as expected
hdul = fits.open(path[0])
hdul.info()
```

```python
import os

# Rename
os.rename(path[0], os.path.join(os.path.dirname(path[0]), file_name))
```
